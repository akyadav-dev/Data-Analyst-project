```python
# for converting from scientific form to decimal form
pop_cols = [c for c in df.columns if 'Population' in c and c != 'World Population Percentage']

df[pop_cols] = df[pop_cols].astype('Int64')   # capital "I" keeps missing values (NaN) safe
```

```python
df.isnull().sum()
```

```python
df.nunique()
```

```python
num_col =df.select_dtypes(include="number").columns
cat_col =df.select_dtypes(include="object").columns

df[num_col]= df[num_col].fillna(df[num_col].median())
df[cat_col]= df[cat_col].fillna("Unknown")
```

```python
df
```

```python
#largest country population wise

df.nlargest(10, '2022 Population')[['Country', '2022 Population']]
```

```python
#smallest country area wise

df.nsmallest(10, 'Area (km²)')[['Country', 'Area (km²)']]
```

```python
#largest country density wise

df.nlargest(10, 'Density (per km²)')[['Country', 'Density (per km²)']]
```

```python
df.groupby('Continent')['2022 Population'].sum().sort_values(ascending=False)
df.groupby('Continent')[['Density (per km²)', 'Growth Rate']].mean()
```

```python
df['Growth_1970_2022 (%)'] = (df['2022 Population'] - df['1970 Population']) / df['1970 Population'] * 100
df['Absolute_Change'] = df['2022 Population'] - df['1970 Population']
```
